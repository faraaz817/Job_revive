```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report
from imblearn.over_sampling import RandomOverSampler

# Step 1: Data Preprocessing
# Assuming 'df' is the loaded dataset
# Perform necessary preprocessing steps

# Step 2: Select Features and Target
X = df.drop('target', axis=1).iloc[:, :304]  # Drop the target variable and select the first 200 columns as features
y = df['target']  # Target variable

# Step 3: Oversample the minority class
oversampler = RandomOverSampler(random_state=150)
X_resampled, y_resampled = oversampler.fit_resample(X, y)

# Step 4: Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(X_resampled, y_resampled, test_size=0.2, random_state=22)

# Step 5: Train the Decision Tree Model
model = DecisionTreeClassifier()
model.fit(X_train, y_train)

# Step 6: Evaluate the Model
y_pred = model.predict(X_test)
print(classification_report(y_test, y_pred))

# Step 7: Custom Prediction
custom_data = [[57,	0,	1,	130,	236	,0	,0	,174	,0	,0	,1,	1	,2]]  # Use a sample from the test set for custom prediction
custom_prediction = model.predict(custom_data)
print("Custom Prediction:", custom_prediction)
```

```python
from sklearn.tree import DecisionTreeClassifier
import pandas as pd

# Assuming 'df' is the loaded dataset
# Step 1: Data Preprocessing
# Perform necessary preprocessing steps

# Step 2: Select Features and Target
X = df.drop('target', axis=1)  # Features
y = df['target']  # Target variable

# Step 3: Train the Decision Tree Model
model = DecisionTreeClassifier()
model.fit(X, y)

# Step 4: Get Feature Importances
importance = model.feature_importances_
feature_names = X.columns

# Step 5: Create a DataFrame to visualize feature importances
importance_df = pd.DataFrame({'Feature': feature_names, 'Importance': importance})
importance_df = importance_df.sort_values('Importance', ascending=False)

# Step 6: Print the feature importances in descending order
print(importance_df)
```

```python
import matplotlib.pyplot as plt

# Plotting the feature importances
plt.figure(figsize=(10, 6))
plt.barh(importance_df['Feature'], importance_df['Importance'])
plt.xlabel('Importance')
plt.ylabel('Feature')
plt.title('Feature Importances')
plt.show()
```
